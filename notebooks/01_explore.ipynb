# ReviewLens - Day 2: Explore the review CSV with pandas

Dataset: Starbucks reviews (Kaggle: harshalhonde/starbucks-reviews-dataset). Run on Kaggle because the local PC blocks compiled Python packages.

**Results:** 705 reviews, 78.0% negative (rating <= 2), average review length 488 characters.

```python
import os
import pandas as pd

# 1. Find the CSV anywhere inside the attached Kaggle dataset
csv_files = []
for root, dirs, files in os.walk("/kaggle/input"):
    for f in files:
        if f.lower().endswith(".csv"):
            csv_files.append(os.path.join(root, f))

if not csv_files:
    raise SystemExit("No CSV found. Add the dataset as an input first.")

path = csv_files[0]
print("Using file:", path)
df = pd.read_csv(path)
print("Columns:", list(df.columns))

# 2. Rating and text columns
rating_col = "Rating"
text_col = "Review"

# 3. Clean: ratings to numbers, drop rows missing a rating or text
df[rating_col] = pd.to_numeric(df[rating_col].astype(str).str.extract(r"(\d+\.?\d*)")[0], errors="coerce")
df = df.dropna(subset=[rating_col, text_col])

# 4. The three Day 2 numbers
total = len(df)
pct_negative = (df[rating_col] <= 2).mean() * 100
avg_len = df[text_col].astype(str).str.len().mean()

print()
print(f"Total reviews: {total}")
print(f"Negative (rating <= 2): {pct_negative:.1f}%")
print(f"Average review length: {avg_len:.0f} characters")

df[[rating_col, text_col]].head()
```
